```python
df_region=pd.read_csv('/kaggle/input/datasets/heesoo37/120-years-of-olympic-history-athletes-and-results/noc_regions.csv')
```

```python
df=pd.read_csv('/kaggle/input/datasets/heesoo37/120-years-of-olympic-history-athletes-and-results/athlete_events.csv')
```

```python
df.head()
```

```python
df=df[df['Season']=='Summer']
```

```python
df.info()
```

```python
df=df.merge(df_region,on='NOC',how='left')
```

```python
df.isnull().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df['Medal'].unique()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder = OneHotEncoder(sparse_output=False).set_output(transform="pandas")
encoded_cols=encoder.fit_transform(df[['Medal']])
df=df.drop(columns=['Medal']).join(encoded_cols)
```

```python
df=df.drop(columns=['Medal_nan'])
```

```python
df
```

```python
df.drop_duplicates(subset=['Team','NOC','Games','Year','City','Sport','Event'],inplace=True)
```

```python
df.info()
```

```python
medal_tally=df.groupby('NOC').sum()[['Medal_Bronze','Medal_Gold','Medal_Silver']].sort_values('Medal_Gold',ascending=False).reset_index()
```

```python
medal_tally
```

```python
country_by_year=df.drop_duplicates(subset=['Year','region'])['Year'].value_counts().reset_index().sort_values('Year')
```

```python
country_by_year
```

```python
import plotly.express as px
```

```python
fig=px.line(country_by_year,x="Year",y="count")
fig.show()
```

```python
events_by_year=df.drop_duplicates(subset=['Year','Event'])['Year'].value_counts().reset_index().sort_values('Year')
```

```python
fig=px.line(events_by_year,x="Year",y="count")
fig.show()
```

```python
athletes_by_year=df.drop_duplicates(subset=['Year','Name'])['Year'].value_counts().reset_index().sort_values('Year')
```
